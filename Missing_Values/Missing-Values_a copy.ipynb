```python
import pandas as pd
import numpy as np
```

```python
customer = pd.read_csv('filse/customer_survey.csv')
customer
```

```python
customer.Gender.unique()
```

```python
customer.SatisfactionScore.unique()
```

```python
customer.Region.unique()
```

```python
customer.Age.unique()
```

```python
customer.replace(("Unknown", -1, "?"), np.nan, inplace=True)
```

```python
cols_to_drop = customer.isnull().sum() / customer.shape[0]  * 100 
cols_to_drop = cols_to_drop[cols_to_drop > 40].index
customer.drop(columns=cols_to_drop , inplace=True)

```

```python
customer.Gender.fillna("Other", inplace=True)
```

```python
customer.Age.fillna(customer.Age.median(), inplace=True)
```

```python
customer.SatisfactionScore.fillna(customer.SatisfactionScore.mode()[0], inplace=True)
```

```python
customer.isnull().sum()
```

```python
customer.to_csv("customer_survey_cleaned_b.csv")
```
